```python
import tensorflow as tf
# Load and preprocess the MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
 
x_train = x_train.astype('float32') / 255.0# scaling min max scaler
x_test = x_test.astype('float32') /255.0
#flatten the images (28*28 into 784)
x_train = x_train.reshape(x_train.shape[0], -1)
x_test = x_test.reshape(x_test.shape[0], -1)
#one - hot encoding labels
y_train = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_test = tf.keras.utils.to_categorical(y_test, num_classes=10)
 
model = tf.keras.Sequential([
    tf.keras.layers.Dense(200, activation = 'rlue', input_shape =(784,)),
    tf.keras.layers.Dense(100, activation = 'sigmoid'),
    tf.keras.layers.Dense(60, activation = 'sigmoid'),
    tf.keras.layers.Dense(30, activation = 'sigmoid'),
    tf.keras.layers.Dense(10, activation = 'softmax')
])
model.summary()
```

```python
# استخدمنا categorical_crossentropy لأن y_train معمولها One-Hot Encoding
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='categorical_crossentropy', 
              metrics=['accuracy'])

# التدريب
history = model.fit(x_train, y_train, 
                    epochs=10, 
                    batch_size=64, 
                    validation_split=0.2)
```

```python
import matplotlib.pyplot as plt

# بنعمل شكل بياني جواه رسمتين جنب بعض، مقاسهم 12 في 4
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. رسم الـ Accuracy (الدقة)
ax1.plot(history.history['accuracy'], label='Training Accuracy', color='blue')
ax1.plot(history.history['val_accuracy'], label='Validation Accuracy', color='orange')
ax1.set_title('Model Accuracy')
ax1.set_xlabel('Epochs')
ax1.set_ylabel('Accuracy')
ax1.legend() # عشان يظهر مربع الدليل
ax1.grid(True) # بيعمل خطوط شبكة عشان القراية تكون أسهل

# 2. رسم الـ Loss (الخطأ)
ax2.plot(history.history['loss'], label='Training Loss', color='blue')
ax2.plot(history.history['val_loss'], label='Validation Loss', color='orange')
ax2.set_title('Model Loss')
ax2.set_xlabel('Epochs')
ax2.set_ylabel('Loss')
ax2.legend()
ax2.grid(True)

# عرض الرسمة
plt.tight_layout()
plt.show()
```

```python
import numpy as np
def get_random_indices(total_images, num_samples=5):
    return np.random.choice(total_images, num_samples, replace=False)
   
```

```python
predictions = model.predict(x_test)

# التعديل هنا: استخدمنا reshape() بدلاً من shape()
# ويفضل نغير اسم المتغير لـ x_test_reshaped عشان يكون أدق
x_test_reshaped = x_test.reshape(-1, 28, 28)

pred_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(y_test, axis=1)
```

```python
import numpy as np
import matplotlib.pyplot as plt

num_images_to_show = 5

# 1. اختيار أرقام (Indices) عشوائية من التست داتا
random_indices = np.random.choice(len(x_test_reshaped), num_images_to_show, replace=False)

# 2. تجهيز الدالة
def plot_images(images, labels, predictions):
    plt.figure(figsize=(15, 4)) # ظبطت المقاس عشان الصور تطلع مرتاحة جنب بعض
    
    # التعديل هنا: استخدام len(images)
    for i in range(len(images)):
        plt.subplot(1, len(images), i+1)
        plt.imshow(images[i], cmap='gray')
        
        # حركة إضافية: لو التوقع صح العنوان يبقى أخضر، لو غلط يبقى أحمر!
        color = 'green' if labels[i] == predictions[i] else 'red'
        
        plt.title(f'Actual: {labels[i]}\nPred: {predictions[i]}', color=color)
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()

# 3. تجميع البيانات العشوائية اللي اخترناها
selected_images = x_test_reshaped[random_indices]
selected_labels = true_classes[random_indices]
selected_preds = pred_classes[random_indices]

# 4. تشغيل الدالة عشان ترسم
plot_images(selected_images, selected_labels, selected_preds)
```

```python
import tensorflow as tf
from tensorflow.keras.callbacks import TensorBoard
 
import datetime

# 1. تحديد المسار اللي هنحفظ فيه السجلات (بنستخدم الوقت عشان كل تجربة تتسجل في فولدر مختلف)
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")

# 2. تعريف الـ TensorBoard Callback
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)

# التعديل: إضافة طبقة Flatten في الأول، وشيلنا الـ input_shape من الـ Dense
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)), # الطبقة دي هتفرد الصور أوتوماتيك
    tf.keras.layers.Dense(200, activation='relu'),
    tf.keras.layers.Dense(100, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# ملحوظة: اتأكد إنك بتستخدم x_train (سمول) لو كنت معرفها كدا فوق، 
# عشان في الـ Error عندك مكتوبة X_train (كابيتال)
history = model.fit(X_train, y_train, 
                    epochs=10, 
                    batch_size=64, 
                    validation_split=0.2, 
                    callbacks=[tensorboard_callback]) 

# التعديل في السطرين دول بس
# ضفنا .reshape(-1, 28, 28) عشان نرجع البيانات لصور مربعة زي ما الموديل متوقع
test_loss, test_acc = model.evaluate(x_test.reshape(-1, 28, 28), y_test)

print(f'Test accuracy: {test_acc}')
```
